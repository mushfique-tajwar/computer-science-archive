# Hash Tables

A Hash Table is a data structure that stores data in an associative manner and fast to retrieve. The data is stored in an array format, where each data value has its own unique index value. Accessing a data value can be done quickly if we know its index. Hashing is used to compute an index into an array of buckets or slots, from which the desired value can be found.

## Creating a Hash Table

```python
hash_table = {
    'key1': 'value1',
    'key2': 'value2',
    'key3': 'value3'
}
```

## Accessing Values

```python
print(hash_table['key1'])
```

## Adding Values

```python
hash_table['key4'] = 'value4'
```

## Updating Values

```python
hash_table['key1'] = 'new_value1'
```

## Removing Values

```python
del hash_table['key2']
```

## Print the Hast Table

```python
print(hash_table)
```

## Common Operations
- **Insert:** Adds a new key-value pair to the hash table.
- **Lookup:** Retrieves the value associated with a given key.
- **Delete:** Removes a key-value pair from the hash table.
- **Update:** Modifies the value associated with a given key.

## Handling Collisions
Collisions occur when two keys hash to the same index. Python's `dict` handles collisions internally using a process called "chaining." When a collision happens, Python stores multiple key-value pairs in the same slot by linking them together in a linked list or another data structure.

## Practical Applications
- **Caching:** Storing data that can be quickly retrieved.
- **Database Indexes:** Speeding up data retrieval in databases.
- **Frequency Counters:** Counting the occurrences of items in a dataset.
- **Cryptography:** Hashing passwords or data for security.
- **Compilers:** Resolving symbol names in code.

## Time Complexity
- **Insert:** Average time complexity is O(1), worst case is O(n).
- **Lookup:** Average time complexity is O(1), worst case is O(n).
- **Delete:** Average time complexity is O(1), worst case is O(n).

## Implementing a Simple Hash Table

```python
class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self._hash(key)
        for kv in self.table[index]:
            if kv[0] == key:
                kv[1] = value
                return
        self.table[index].append([key, value])

    def get(self, key):
        index = self._hash(key)
        for kv in self.table[index]:
            if kv[0] == key:
                return kv[1]
        raise KeyError(f"Key '{key}' not found")

    def delete(self, key):
        index = self._hash(key)
        for i, kv in enumerate(self.table[index]):
            if kv[0] == key:
                del self.table[index][i]
                return
        raise KeyError(f"Key '{key}' not found")

# Example usage
ht = HashTable()
ht.insert('name', 'Alice')
ht.insert('age', 30)
print(ht)
print(ht.get('name'))
ht.delete('age')
```
