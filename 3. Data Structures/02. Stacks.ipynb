# Stacks
Stacks are a linear data structure that follows the Last In First Out (LIFO) principle. They are particularly useful for tasks that involve managing temporary data, such as parsing expressions, implementing function calls, and managing recursive processes.

## Basic Concepts
- **Stack:** A stack is a collection of elements that follows the Last In First Out (LIFO) principle.
- **Push:** Adding an element to the top of the stack.
- **Pop:** Removing the element from the top of the stack.
- **Peek/Top:** Accessing the top element of the stack without removing it.
- **IsEmpty:** Checking if the stack is empty.
- **Size:** Getting the number of elements in the stack.

## Implementing a Stack

````python
class Stack:
    def __init__(self):
        self.items = []
    
    def is_empty(self):
        return len(self.items) == 0
    
    def push(self, item):
        self.items.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None
    
    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return None
    
    def size(self):
        return len(self.items)
````

## Using the Stack Class

````python
# Create a stack
stack = Stack()

# Push elements onto the stack
stack.push(1)
stack.push(2)
stack.push(3)

# Pop elements from the stack
print("Element popped:",stack.pop())
print("Element popped:",stack.pop())

# Peek at the top element
print("Top element:",stack.peek())

# Check if the stack is empty
print("Is stack empty?\n=",stack.is_empty())

# Get the size of the stack
print("Size of stack:",stack.size())
````

## Applications of Stacks
- **Function Call Stack:** In languages like C, Java, and Python, function calls are managed using a stack.
- **Expression Evaluation:** Parsing and evaluating expressions like arithmetic or boolean expressions.
- **Backtracking Algorithms:** Algorithms that explore all possible solutions and backtrack when a solution is not possible.
- **Binary Conversion:** Converting numbers between different bases (binary, octal, decimal, hexadecimal).
- **Balancing Symbols:** Checking if parentheses, brackets, and braces are balanced in an expression.

## Example
Balancing parenthesis

````python
def is_balanced(expression):
    stack = Stack()
    matching_bracket = {')': '(', '}': '{', ']': '['}
    
    for char in expression:
        if char in matching_bracket.values():
            stack.push(char)
        elif char in matching_bracket:
            if stack.is_empty() or stack.pop() != matching_bracket[char]:
                return False
        else:
            return False
    
    return stack.is_empty()

# Test cases
print("Is '()' Balanced:",is_balanced("()"))
print("Is '()[]{}' Balanced:",is_balanced("()[]{}"))
print("Is '(]' Balanced:",is_balanced("(]"))
print("Is '([)]' Balanced:",is_balanced("([)]"))
print("Is '{[]}' Balanced:",is_balanced("{[]}"))
print("Is '[()]' Balanced:",is_balanced("[()]"))
````

## Advanced Topics
- **Stacks with Exceptions:** You can raise exceptions when operations like `pop` or `peek` are called on an empty stack.
- **Efficiency Considerations:** The time complexity for `push`, `pop`, and `peek` operations in a stack implemented with a list is O(1). However, if the stack grows beyond the current capacity, it may need to resize, leading to O(n) complexity.
- **Using Python’s `collections.deque`:** Python’s `collections.deque` provides an efficient implementation of a stack with O(1) time complexity for the `append` (push) and ```pop``` operations.

````python
from collections import deque

class Stack:
    def __init__(self):
        self.items = deque()
    
    def is_empty(self):
        return len(self.items) == 0
    
    def push(self, item):
        self.items.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None
    
    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return None
    
    def size(self):
        return len(self.items)
````
