# Queues

Queues are a fundamental data structure in computer science that follows the First In First Out (FIFO) principle. They are useful for managing tasks that require a sequence of operations, such as task scheduling, breadth-first search (BFS), and managing limited resources.

## Basic Concepts
- **Queue:** A queue is a collection of elements that follows the First In First Out (FIFO) principle.
- **Enqueue:** Adding an element to the rear of the queue.
- **Dequeue:** Removing the element from the front of the queue.
- **Front:** Accessing the front element of the queue without removing it.
- **Rear:** Accessing the rear element of the queue without removing it.
- **IsEmpty:** Checking if the queue is empty.
- **Size:** Getting the number of elements in the queue.

## Implementing a Queue

```python
class Queue:
    def __init__(self):
        self.items = []
    
    def is_empty(self):
        return len(self.items) == 0
    
    def enqueue(self, item):
        self.items.append(item)
    
    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        return None
    
    def front(self):
        if not self.is_empty():
            return self.items[0]
        return None
    
    def rear(self):
        if not self.is_empty():
            return self.items[-1]
        return None
    
    def size(self):
        return len(self.items)
```

## Using the Queue Class

```python
# Create a queue
queue = Queue()

# Enqueue elements into the queue
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)

# Dequeue elements from the queue
print('Element dequeued:', queue.dequeue())
print('Element dequeued:', queue.dequeue())

# Access the front and rear elements
print('Front element:', queue.front())
print('Rear element:', queue.rear())

# Check if the queue is empty
print('Is queue empty?', queue.is_empty())

# Get the size of the queue
print('Size of queue:', queue.size())
```

## Applications of Queues
- **Task Scheduling:** Managing tasks that need to be executed in the order they arrive.
- **Breadth-First Search (BFS):** A graph traversal algorithm that explores nodes level by level.
- **Printing Systems:** Managing print jobs in a printer.
- **CPU Scheduling:** Managing CPU processes in an operating system.
- **Buffering:** Storing data temporarily before processing.

## Example
Breadth-First Search (BFS)

```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    traversal_order = []
    
    while queue:
        node = queue.popleft()
        traversal_order.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return traversal_order

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print(bfs(graph, 'A'))
```

## Advanced Topics
- **Priority Queues:** A queue where each element is associated with a priority, and the elements are served based on their priority.
- **Circular Queues:** A queue where the last element points back to the first element, making it a circular structure.
- **Using Python’s `collections.deque`:** Python’s `collections.deque` provides an efficient implementation of a queue with O(1) time complexity for the `append` (enqueue) and `popleft` (dequeue) operations.

```python
from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()
    
    def is_empty(self):
        return len(self.items) == 0
    
    def enqueue(self, item):
        self.items.append(item)
    
    def dequeue(self):
        if not self.is_empty():
            return self.items.popleft()
        return None
    
    def front(self):
        if not self.is_empty():
            return self.items[0]
        return None
    
    def rear(self):
        if not self.is_empty():
            return self.items[-1]
        return None
    
    def size(self):
        return len(self.items)
```
