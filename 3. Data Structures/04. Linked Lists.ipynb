# Linked Lists
A linked list is a linear data structure where each element (node) contains a reference (or link) to the next node. Unlike arrays, where elements are stored in contiguous memory locations, linked lists store elements in scattered memory locations and each element is linked to the next using pointers or references.

### Types of Linked Lists
- **Singly Linked List:** Each node contains a data part and a reference (link) to the next node.
- **Doubly Linked List:** Each node contains data part and references to both the next and the previous node.
- **Circular Linked List:** The last node points back to the head, forming a circular loop.
### Basic Components
- **Node:** The basic building block of a linked list. Each node contains data and a reference to the next node.
- **Head:** The first node in the list.
- **Tail:** The last node in the list.
- **Traversal:** The process of moving through each node in the list, typically from the head to the tail.

## Implementing a Singly Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def append(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def prepend(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def delete(self, data):
        current = self.head
        previous = None
        while current:
            if current.data == data:
                if previous:
                    previous.next = current.next
                else:
                    self.head = current.next
                return
            previous = current
            current = current.next

    def search(self, data):
        current = self.head
        while current:
            if current.data == data:
                return True
            current = current.next
        return False

    def traverse(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
```

```python
my_sll = LinkedList()
print("Is the linked list empty?", my_sll.is_empty())
print()
print("Appending 1, 2, 3 to the linked list.\n")
my_sll.append(1)
my_sll.append(2)
my_sll.append(3)
print("Is the linked list empty?", my_sll.is_empty())
print()
print("Traversing the linked list:")
my_sll.traverse()

print()
print("Prepending 0 to the linked list.\n")

my_sll.prepend(0)
print("Traversing the linked list:")
my_sll.traverse()

print()
print("Deleting 2 the linked list.\n")
my_sll.delete(2)
print("Traversing the linked list:")
my_sll.traverse()

print("Do 1 exist in the linked list?", my_sll.search(1))
print("Do 2 exist in the linked list?", my_sll.search(2))
```

## Doubly Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None
```

```python
class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def is_empty(self):
        return self.head is None

    def append(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.head = self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    def prepend(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.head = self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    def delete(self, data):
        current = self.head
        while current:
            if current.data == data:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next
                if current.next:
                    current.next.prev = current.prev
                else:
                    self.tail = current.prev
                return
            current = current.next

    def search(self, data):
        current = self.head
        while current:
            if current.data == data:
                return True
            current = current.next
        return False

    def traverse(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
```

```python
my_dll = DoublyLinkedList()
print("Is the doubly linked list empty?", my_dll.is_empty())
print()
print("Appending 1, 2, 3 to the doubly linked list.\n")
my_dll.append(1)
my_dll.append(2)
my_dll.append(3)
print("Traversing the doubly linked list:")
my_dll.traverse()


print()
print("Prepending 0 to the doubly linked list.\n")
my_dll.prepend(0)
print("Traversing the doubly linked list:")
my_dll.traverse()

print()
print("Deleting 2 the doubly linked list.\n")
my_dll.delete(2)
print("Traversing the doubly linked list:")
my_dll.traverse()

print()
print("Searching for 1 and 2 in the doubly linked list:")
print(my_dll.search(1))
print(my_dll.search(2))
```

## Circular Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class CircularLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def append(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            new_node.next = self.head
        else:
            current = self.head
            while current.next != self.head:
                current = current.next
            current.next = new_node
            new_node.next = self.head

    def prepend(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            new_node.next = self.head
        else:
            current = self.head
            while current.next != self.head:
                current = current.next
            current.next = new_node
            new_node.next = self.head
            self.head = new_node

    def delete(self, data):
        if self.is_empty():
            return

        # Single node
        if self.head.next == self.head:
            if self.head.data == data:
                self.head = None
            return

        # Deleting the head: find the tail so it can be re-linked
        if self.head.data == data:
            tail = self.head
            while tail.next != self.head:
                tail = tail.next
            self.head = self.head.next
            tail.next = self.head
            return

        # Deleting any other node
        previous = self.head
        current = self.head.next
        while current != self.head:
            if current.data == data:
                previous.next = current.next
                return
            previous = current
            current = current.next

    def search(self, data):
        current = self.head
        while current:
            if current.data == data:
                return True
            current = current.next
            if current == self.head:
                break
        return False

    def traverse(self):
        if self.is_empty():
            print("List is empty")
            return
        current = self.head
        while True:
            print(current.data, end=" -> ")
            current = current.next
            if current == self.head:
                break
        print("None")
```

```python
my_cll = CircularLinkedList()
print("Is the circular linked list empty?", my_cll.is_empty())
print()
print("Append some elements to the circular linked list")
my_cll.append(1)
my_cll.append(2)
my_cll.append(3)

print()
print("Traversing the circular linked list:")
my_cll.traverse()

print()
print("Prepending 0 to the circular linked list:")
my_cll.prepend(0)
print()
print("Traversing the circular linked list:")
my_cll.traverse()

print()
print("Deleting 2 from the circular linked list:")
my_cll.delete(2)
print()
print("Traversing the circular linked list:")
my_cll.traverse()

print()
print("Searching for 1 and 2 in the circular linked list:")
print(my_cll.search(1))  # Output: True
print(my_cll.search(2))  # Output: False
```

## Edge Cases
- **Empty List:** When the list is empty, operations like delete and search need to handle this case.
- **Deleting the Head:** When the node to delete is the head of the list.
Deleting the Tail: When the node to delete is the tail of the list.
## Advantages and Disadvantages
### Advantages:

- **Dynamic Size:** Can grow and shrink in size dynamically.
- **Insertions and Deletions:** Insertions and deletions can be done in constant time at the head of the list.
### Disadvantages:

- **Random Access:** No direct access to any element. You need to traverse the list to reach a particular element.
- **Extra Memory:** Each node requires extra memory for storing the pointer to the next node.
## Time Complexity
- **Access:** O(n)
- **Insertion:** O(1) at the head, O(n) at the end
- **Deletion:** O(1) at the head, O(n) at the end
## Space Complexity
O(n), where n is the number of nodes in the list.
## Practical Use Cases
- **Stacks and Queues:** Implementing these data structures using linked lists.
- **Implementing Custom Data Structures:** For example, a priority queue or a custom stack.
- **Dynamic Data Storage:** When the number of elements is not known beforehand.
