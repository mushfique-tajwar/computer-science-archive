# Arrays

An array is a data structure that stores a collection of elements of the same data type. These elements are stored in contiguous memory locations. Each element in the array has a unique index, which starts from 0 (in many programming languages) and goes up to the total number of elements minus one.

### Purpose of Arrays

- **Efficient Data Storage:** Arrays allow for efficient storage of large amounts of data.
- **Easy Access:** Elements in an array can be accessed directly using their index, making it faster to retrieve elements compared to other data structures like linked lists.
- **Efficient Manipulation:** Operations such as adding, deleting, and updating elements can be performed efficiently using array indices.

### Pros of Arrays

- **Fast Access:** Arrays provide fast access to elements using their index.
- **Efficient Memory Usage:** Arrays use memory efficiently as they store elements in contiguous memory locations.
- **Easy to Traverse:** Iterating over all elements in an array is straightforward.

### Cons of Arrays

- **Fixed Size:** Arrays have a fixed size, meaning you must know the size of the array at the time of creation. If you need to add more elements, you need to create a new array with a larger size and copy the elements from the old array to the new one.
- **Memory Wasted:** If you create an array with a size that is much larger than the number of elements you need, you are wasting memory.
- **Insertions and Deletions:** Inserting or deleting elements in the middle of an array can be costly as it may require shifting elements to maintain the order of the array.

### Use Cases

- **Data Storage:** Storing a list of items, such as student scores, product prices, etc.
- **Sorting and Searching:** Implementing sorting algorithms like bubble sort, insertion sort, and searching algorithms like linear search and binary search.
- **Matrix Operations:** Performing operations on matrices, such as addition, multiplication, and transposition.
- **Image Processing:** Manipulating image data, such as pixel values.
- **Game Development:** Storing game state information, such as player positions and scores.

## Importing Numpy

```python
import numpy as np
```

## Creating Arrays
You can create a NumPy array using the np.array() function.

```python
arr1 = np.array([0]*5)
print(arr1)
print(type(arr1), "\n")

arr2 = np.array([1, 2, 3, 4])
print(arr2)
print(type(arr2), "\n")

arr3 = np.zeros(6, dtype=int)
print(arr3)
print(type(arr3))
```

## Specifying the Data Type
You can specify the data type of the array elements.

```python
# Creating an array with integar data type
c = np.array([1, 2, 3], dtype=int)
print(c, "\n")

# Creating an array with float data type
c = np.array([1, 2, 3], dtype=float)
print(c, "\n")

# Creating an array with string data type
c = np.array([1, 2, 3], dtype=str)
print(c)
```

## Generating Arrays
NumPy provides functions to generate arrays with specific patterns.

```python
# Creating an array of zeros
zeros = np.zeros((2, 3))
print(zeros, "\n")

# Creating an array of ones
ones = np.ones((2, 3))
print(ones, "\n")

# Creating an array with a range of values
arange = np.arange(0, 10, 2)
print(arange)
```

## Multi-Dimensional Arrays

### 2D Array

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
```

### 3D Array

```python
b = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(b)
```

### Accessing Elements

```python
# Accessing elements in a 2D array
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a[0, 1])

# Accessing elements in a 3D array
b = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(b[1, 0, 1])
```

### Slicing

```python
# Slicing a 2D array
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a[0, 1:])

# Slicing a 3D array
b = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(b[0, :, 1])
```

## Array Operations

### Basic Arithmetic

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)
print(a * b)
```

### Universal Functions (ufuncs)
NumPy provides universal functions to perform element-wise operations.

```python
a = np.array([1, 2, 3])
print(np.sin(a))
```

### Broadcasting
Broadcasting allows arrays of different shapes to be combined in arithmetic operations.

```python
a = np.array([1, 2, 3])
b = np.array([4])
print(a + b)
```

## Array Reshaping

### Reshaping

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape((2, 3))
print(b)
```

### Flatten

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = a.flatten()
print(b)
```

## Array Manipulation

### Transposing

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = a.transpose()
print(b)
```

### Flipping

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.flip(a, axis=0)
print(b)
```
