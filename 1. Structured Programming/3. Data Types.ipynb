### 1. Numeric Data Types

```python
var1 = 1       # int data type
var2 = True    # bool data type
var3 = 10.023  # float data type
var4 = 10+3j   # complex data type

print(var1, type(var1))
print(var2, type(var2))
print(var3, type(var3))
print(var4, type(var4))
```

### 2. String Data Types

```python
var5 = "Rutab"  # str data type
print(var5, type(var5))
```

### 3. Sequence Data Types

3.1 List

```python
list = [ 'abcd', 786 , 2.23, 'rutab', 70.2 ]
tinylist = [123, 'rutab']

print (list)            # Prints complete list
print (list[0])         # Prints first element of the list
print (list[1:3])       # Prints elements starting from 2nd till 3rd 
print (list[2:])        # Prints elements starting from 3rd element
print (tinylist * 2)    # Prints list two times
print (list + tinylist) # Prints concatenated lists
```

3.2 Tuple

```python
tuple = ( 'abcd', 786 , 2.23, 'rutab', 70.2  )
tinytuple = (123, 'rutab')

print (tuple)               # Prints the complete tuple
print (tuple[0])            # Prints first element of the tuple
print (tuple[1:3])          # Prints elements of the tuple starting from 2nd till 3rd 
print (tuple[2:])           # Prints elements of the tuple starting from 3rd element
print (tinytuple * 2)       # Prints the contents of the tuple twice
print (tuple + tinytuple)   # Prints concatenated tuples
```

3.3 Range

`range(start, stop, step)`

```python
for i in range(1, 10, 2):
  print(i)
```

### 4. Binary Data Types

4.1 Bytes

Based on ASCII

```python
b1 = bytes([65, 66, 67, 68, 69])  
print(b1)
print('----')
b2 = b'Hello'  
print(b2)  
```

4.2 Bytearray

Also based on ASCII

```python
value = bytearray([72, 101, 108, 108, 111])  
print(value)  
print('----')
val = bytearray("Hello", 'utf-8')  
print(val)  
```

4.3 Memoryview

```python
data = bytearray(b'Hello, world!')
view = memoryview(data)
print(view)
```

```python
# For array

import array
arr = array.array('i', [1, 2, 3, 4, 5])
view = memoryview(arr)
print(view)
```

```python
# For a certain part of a data

data = b'Hello, world!'
# Creating a view of the last part of the data
view = memoryview(data[7:])  
print(view)
```

### 5. Dictionary Data Types

Similar to hash table

```python
dict = {}
dict['one'] = "This is one"
dict[2]     = "This is two"

tinydict = {'name': 'Rutab','code':6734, 'dept': 'Development'}


print (dict['one'])       # Prints value for 'one' key
print (dict[2])           # Prints value for 2 key
print (tinydict)          # Prints complete dictionary
print (tinydict.keys())   # Prints all the keys
print (tinydict.values()) # Prints all the values
```

### 6. Set Data Types

```python
set1 = {123, 452, 5, 6}
set2 = {'Java', 'Python', 'JavaScript'}

print(set1)
print(set2)
```

### 7. Boolean Data Types

```python
# Returns false as a is not equal to b
a = 2
b = 4
print(bool(a==b))

# Following also prints the same
print(a==b)

# Returns False as a is None
a = None
print(bool(a))

# Returns false as a is an empty sequence
a = ()
print(bool(a))

# Returns false as a is 0
a = 0.0
print(bool(a))

# Returns false as a is 10
a = 10
print(bool(a))
```

### 8. None Type

```python
# Declaring a variable
# And, assigning a Null value (None)

x = None

# Printing its value and type
print("x = ", x)
print("type of x = ", type(x))
```

## Primitive and Non-primitive Data Types


The above-explained data types can also be categorized as primitive and non-primitive.

**1. Primitive Types:** The primitive data types are the fundamental data types that are used to create complex data types (sometimes called complex data structures). There are mainly four primitive data types, which are:

- Integers
- Floats
- Booleans
- Strings

**2. Non-primitive Types:** The non-primitive data types store values or collections of values. There are mainly four types of non-primitive types, which are:

- Lists
- Tuples
- Dictionaries
- Sets

## Data Type Conversion

```python
print("Conversion to integer data type")
a = int(1)     # a will be 1
b = int(2.2)   # b will be 2
c = int("3")   # c will be 3

print (a)
print (b)
print (c)

print("Conversion to floating point number")
a = float(1)     # a will be 1.0
b = float(2.2)   # b will be 2.2
c = float("3.3") # c will be 3.3

print (a)
print (b)
print (c)

print("Conversion to string")
a = str(1)     # a will be "1" 
b = str(2.2)   # b will be "2.2"
c = str("3.3") # c will be "3.3"

print (a)
print (b)
print (c)
```

`int()`:        Converts x to an integer. base specifies the base if x is a string.

`long()`:       Converts x to a long integer. base specifies the base if x is a string. This function has been deprecated.

`float()`:      Converts x to a floating-point number.

`complex()`:    Creates a complex number.

`str()`:        Converts object x to a string representation.

`repr()`:       Converts object x to an expression string.

`eval()`:       Evaluates a string and returns an object.

`tuple()`:      Converts s to a tuple.

`list()`:       Converts s to a list.

`set()`:        Converts s to a set.

`dict()`:       Creates a dictionary. d must be a sequence of (key,value) tuples.

`frozenset()`:  Converts s to a frozen set.

`chr()`:        Converts an integer to a character.

`unichr()`:     Converts an integer to a Unicode character.

`ord()`:        Converts a single character to its integer value.

`hex()`:        Converts an integer to a hexadecimal string.

`oct()`:        Converts an integer to an octal string.
