### If-Else

```python
var = 100
if (var == 100): 
   print ("Value of var is equal to 100")
else:
   print("Value of var is not equal to 100")
```

### If-Elif-Else

```python
var = 100
if (var > 100): 
   print ("Value of var more than 100")
elif (var == 100): 
   print ("Value of var is equal to 100")
else:
   print("Value of var is less than 100")
```

### Nested

```python
var = int(input("Enter a number: "))
if (var == 100):
    print("Value of var is equal to 100")
else:
    if (var > 100):
        print("Value of var is more than 100")
    else:
        print("Value of var is less than 100")
```

This is also an example of conditional user input
