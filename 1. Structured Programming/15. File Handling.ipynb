# File Handling

## File Opening Modes

| Method | Description |
|---|---|
| `r` | Opens a file for reading only.  |
| `rb` | Opens a file for reading only in binary format. |
| `r+` | Opens a file for both reading and writing. |
| `rb+` | Opens a file for both reading and writing in binary format. |
| `w` | Opens a file for writing only. |
| `b` | Opens the file in binary mode. |
| `t` | Opens the file in text mode (default). |
| `+` | Opens file for updating (reading and writing). |
| `wb` | Opens a file for writing only in binary format. |
| `w+` | Opens a file for both writing and reading. |
| `wb+` | Opens a file for both writing and reading in binary format. |
| `a` | Opens a file for appending. |
| `ab` | Opens a file for appending in binary format. |
| `a+` | Opens a file for both appending and reading. |
| `ab+` | Opens a file for both appending and reading in binary format. |
| `x` | Opens for exclusive creation, failing if the file already exists. |

```python
# Open a file
fo = open("15. File2.txt", "wb")
print ("Name of the file: ", fo.name)
print ("Closed or not: ", fo.closed)
print ("Opening mode: ", fo.mode)
fo.close()
```

## Reading Files

```python
with open("15. File1.txt", "r") as file:
   content = file.read()
   print(content)
```

```python
with open("15. File1.txt", "r") as file:
   line = file.readline()
   while line:
      print(line, end='')
      line = file.readline()
```

```python
with open("15. File1.txt", "r") as file:
   lines = file.readlines()
   for line in lines:
      print(line, end='')
```

## Writing Files

```python
file = open("15. File3.txt", "w")
file.write("Hello, World!")
file.close()
print ("File opened successfully!!")
```

```python
file = open("15. File3.txt", "a")
file.write("Appending this line.\n")
file.close()
print ("File opened successfully!!")
```

```python
# Open a file in write mode
with open("15. File3.txt", "w") as file:
   file.write("Hello, World!\n")
   file.write("This is a new line.\n")
print ("File opened successfully!!")
```

```python
# List of lines to write to the file
lines = ["First line\n", "Second line\n", "Third line\n"]

# Open a file in write mode
with open("15. File3.txt", "w") as file:
    file.writelines(lines)

print ("File opened successfully!!")
```

```python
fo = open("15. File4.txt", "w")
fo.write( "Python is a great language.\nYeah its great!!\n")

# Close opened file
fo.close()
```

```python
# Open a file in binary write mode
with open('15. File5.bin', 'wb') as f:
   # Binary data
   data = b"Hello World"  
   f.write(data)
```

```python
# Open a file in binary write mode
with open('15. File5.bin', 'wb') as f:
   # Convert text string to bytes
   data = "Hello World".encode('utf-8')
   f.write(data)
```

```python
# Open a file in append mode
fo = open("15. File4.txt", "a")
text = "A fabulous Python tutorial"
fo.write(text)

# Close opened file
fo.close()
```

```python
# Open a file in read-write mode
fo = open("15. File8.txt", "w+")

# Write initial data to the file
fo.write("This is a rat race")

# Move the read/write pointer to the 10th byte
fo.seek(10, 0)

# Read 3 bytes from the current position
data = fo.read(3)

# Move the read/write pointer back to the 10th byte
fo.seek(10, 0)

# Overwrite the existing content with new text
fo.write('cat')

# Close the file
fo.close()
```

## Read Files

```python
# Open the file in read mode
file = open('15. File1.txt', 'r')

# Read the entire content of the file
content = file.read()

# Print the content
print(content)

# Close the file
file.close()
```

```python
# Open the file in read mode
file = open('15. File1.txt', 'r')

# Read the first line of the file
line = file.readline()

# Print the line
print(line)

# Close the file
file.close()
```

```python
# Open the file in read mode
file = open('15. File1.txt', 'r')

# Read all lines from the file
lines = file.readlines()

# Print the lines
for line in lines:
   print(line, end='')

# Close the file
file.close()
```

```python
# Using the with statement to open a file
with open('15. File1.txt', 'r') as file:
   content = file.read()
   print(content)
```

```python
# Open the file in binary write mode
with open('15. File5.bin', 'wb') as f:
   data = b"Hello World"
   f.write(data)
```

```python
# Open the file in binary read mode
with open('15. File5.bin', 'rb') as f:
   data = f.read()
   print(data.decode(encoding='utf-8'))
```

```python
# Convert the integer to bytes and write to a binary file
n = 25
data = n.to_bytes(8, 'big')

with open('15. File6.bin', 'wb') as f:
   f.write(data)
```

```python
# Read the binary data from the file and convert it back to an integer
with open('15. File6.bin', 'rb') as f:
   data = f.read()
   n = int.from_bytes(data, 'big')
   print(n)
```

```python
import struct

# Define a floating-point number
x = 23.50

# Pack the float into a binary format
data = struct.pack('f', x)

# Open the file in binary write mode and write the packed data
with open('15. File7.bin', 'wb') as f:
   f.write(data)
```

```python
# Open the file in binary read mode
with open('15. File7.bin', 'rb') as f:
   # Read the binary data from the file
   data = f.read()
    
   # Unpack the binary data to retrieve the float
   x = struct.unpack('f', data)[0]
    
   # Print the float value
   print(x)
```

```python
# Open the file in read-write mode
with open("15. File4.txt", "r+") as fo:
   # Write data to the file
   fo.write("This is a rat race")

   # Rewind the pointer to the beginning of the file
   fo.seek(0)

   # Read data from the file
   data = fo.read()
   print(data)
```

```python
# Open a file in read-write mode
fo = open("15. File8.txt", "w+")

# Write initial data to the file
fo.write("This is a rat race")

# Seek to a specific position in the file
fo.seek(10, 0)

# Read a few bytes from the current position
data = fo.read(3)
print("Data read from position 10:", data)

# Seek back to the same position
fo.seek(10, 0)

# Overwrite the earlier contents with new text
fo.write("cat")

# Rewind to the beginning of the file
fo.seek(0, 0)

# Read the entire file content
data = fo.read()
print("Updated file content:", data)

# Close the file
fo.close()
```

## File Methods

| Method | Description |
|---|---|
| `file.close()` | Close the file. A closed file cannot be read or written any more. |
| `file.flush()` | Flush the internal buffer. |
| `file.fileno()` | Returns the integer file descriptor that is used by the underlying implementation to request I/O operations from the operating system. |
| `file.isatty()` | Returns True if the file is connected to a tty(-like) device, else False. |
| `file.next()` | Returns the next line from the file each time it is being called. |
| `buffer_info()` | Returns a tuple (address, length) giving the current memory address and the length in elements of the buffer used to hold the arrays contents. |
| `file.read([size])` | Reads at most size bytes from the file (less if the read hits EOF before obtaining size bytes). |
| `file.readline([size])` | Reads one entire line from the file. A trailing newline character is kept in the string. |
| `file.readlines([sizehint])` | Reads until EOF using readline() and return a list containing the lines. |
| `file.seek(offset[, whence])` | Sets the file's current position. |
| `file.tell()` | Returns the file's current position. |
| `file.truncate([size])` | Truncates the file's size. |
| `file.write(str)` | Writes a string to the file. There is no return value. |
| `file.writelines(sequence)` | Writes a sequence of strings to the file. |
